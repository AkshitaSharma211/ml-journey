```python
w = torch.tensor(3.0, requires_grad=True)
loss = w ** 2   # pretend this is our loss - we want to minimize it

loss.backward()
print("Gradient:", w.grad)

learning_rate = 0.1
with torch.no_grad():
    w -= learning_rate * w.grad

print("New w:", w)
```

```python
w = torch.tensor(3.0, requires_grad=True)
learning_rate = 0.1

for step in range(10):
    loss = w ** 2
    loss.backward()

    with torch.no_grad():
        w -= learning_rate * w.grad
        w.grad.zero_()   # reset gradient before next step

    print(f"Step {step}: w = {w.item():.4f}, loss = {loss.item():.4f}")
```
