```python
import torch

x = torch.linspace(0, 10, 50)
y_true = 3*x + 5 + torch.randn(50)*2   # true line: y = 3x + 5, plus noise
```

```python
w = torch.randn(1, requires_grad=True)
b = torch.randn(1, requires_grad=True)
```

```python
learning_rate = 0.01   # you need this defined - missing in your attempt

for step in range(1000):   # you need the loop wrapper - missing in your attempt
    y_pred = w * x + b                          # ✓ you got this right
    loss = ((y_pred - y_true) ** 2).mean()       # ✓ you got this right
    loss.backward()                              # ✓ you got this right

    with torch.no_grad():
        w -= learning_rate * w.grad              # ✓ you had this
        b -= learning_rate * b.grad               # ✗ missing - b needs updating too!
        w.grad.zero_()                            # ✓ you had this
        b.grad.zero_()                             # ✗ missing - b's gradient needs resetting too!

    if step % 10 == 0:
        print(f"Step {step}: w = {w.item():.4f}, b = {b.item():.4f}, loss = {loss.item():.4f}")
```
